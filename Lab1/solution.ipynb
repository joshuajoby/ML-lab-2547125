# Lab 1
Data Cleaning and Exploratory Data Analysis

Task 1:
I first loaded both datasets and checked their dimensions, data types, missing values, and summary statistics. This helped me understand the structure of the datasets and identify potential data quality issues before doing any cleaning.

```python
import pandas as pd
import numpy as np

aqi = pd.read_csv("city_day.csv")
crop = pd.read_csv("crop_production.csv")

display(aqi)
display(crop)
print(aqi.shape)
print(crop.shape)

aqi.info()
crop.info()

print(aqi.isnull().sum())
print(crop.isnull().sum())

display(aqi.describe())
display(crop.describe())
```

Observation:
The AQI dataset has 29,531 rows and 16 columns. It covers daily air quality readings across Indian cities from 2015 to 2020. Several columns have missing values, with Xylene missing the most entries (over 18,000). The crop dataset has 246,091 rows and 7 columns. Only the Production column has missing values, around 3,730 entries. AQI values range from 13 to 2,049, which already suggests there are some very extreme readings in the data.

Task 2:
I checked the missing values in both datasets and filled them using appropriate methods. Median imputation was used for numerical columns because it is less affected by extreme values. For the AQI_Bucket column, I used the most frequent category since it is categorical.

```python
aqi_clean = aqi.copy()

for col in aqi_clean.select_dtypes(include="number").columns:
    aqi_clean[col] = aqi_clean[col].fillna(aqi_clean[col].median())

aqi_clean["AQI_Bucket"] = aqi_clean["AQI_Bucket"].fillna(
    aqi_clean["AQI_Bucket"].mode()[0]
)

crop_clean = crop.copy()

crop_clean["Production"] = crop_clean["Production"].fillna(
    crop_clean["Production"].median()
)

print(aqi_clean.isnull().sum())
print(crop_clean.isnull().sum())
```

Observation:
All missing values in both datasets are now filled. Every column shows zero null counts. I chose the median over the mean because some columns like AQI have extreme values that would pull the mean too high and give a misleading fill.

Task 3:
I cleaned the text columns and checked for duplicate records. I removed unnecessary spaces from string columns and dropped exact duplicate rows to make sure the data is consistent before any analysis.

```python
aqi_before = len(aqi_clean)
crop_before = len(crop_clean)

for col in aqi_clean.select_dtypes(include="str").columns:
    aqi_clean[col] = aqi_clean[col].str.strip()

for col in crop_clean.select_dtypes(include="str").columns:
    crop_clean[col] = crop_clean[col].str.strip()

aqi_clean.drop_duplicates(inplace=True)
crop_clean.drop_duplicates(inplace=True)

print("AQI Before:", aqi_before)
print("AQI After:", len(aqi_clean))

print("Crop Before:", crop_before)
print("Crop After:", len(crop_clean))
```

Observation:
Extra spaces were removed from text columns and duplicate records were eliminated to improve data quality. The row counts stayed the same for both datasets, which means there were no exact duplicate rows.

Task 4:
The board wants to know where most AQI values cluster and whether any extreme readings exist. I used a histogram to show where values concentrate and a boxplot to reveal any extreme values. A histogram shows the shape and spread of the data clearly, while a boxplot highlights outliers as individual points beyond the whiskers. Together they answer both questions.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6, 4))
sns.histplot(aqi_clean["AQI"], bins=50, kde=True, color="steelblue")
plt.title("Distribution of AQI Values")
plt.xlabel("AQI")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 3))
sns.boxplot(x=aqi_clean["AQI"], color="salmon")
plt.title("AQI Boxplot")
plt.xlabel("AQI")
plt.tight_layout()
plt.show()
```

Observations:
1. Most AQI values are concentrated between roughly 50 and 250. The distribution is right-skewed, meaning most city-days fall in the moderate pollution range and very high readings are rare.
2. The boxplot shows a number of extreme AQI values above 500, with some going past 1000. These few outliers will pull the average upward and make pollution seem worse than it really is for most places.

Task 5:
I used the IQR method to identify outliers because it works well even when the data is skewed and does not assume a normal distribution. Rather than deleting the extreme rows, I capped the values at the IQR upper boundary. This keeps all the other information in those rows intact while removing the distortion caused by the extreme AQI numbers.

```python
Q1 = aqi_clean["AQI"].quantile(0.25)
Q3 = aqi_clean["AQI"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = aqi_clean[(aqi_clean["AQI"] < lower) | (aqi_clean["AQI"] > upper)]
print(f"Outliers detected: {len(outliers)}")
print(f"Upper cap: {upper:.2f}")

aqi_before_cap = aqi_clean["AQI"].copy()
aqi_clean["AQI"] = aqi_clean["AQI"].clip(lower=lower, upper=upper)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(aqi_before_cap, bins=50, kde=True, color="salmon", ax=axes[0])
axes[0].set_title("AQI Before Treatment")
axes[0].set_xlabel("AQI")
axes[0].set_ylabel("Frequency")

sns.histplot(aqi_clean["AQI"], bins=50, kde=True, color="steelblue", ax=axes[1])
axes[1].set_title("AQI After Capping")
axes[1].set_xlabel("AQI")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()

print("Mean AQI before:", round(aqi_before_cap.mean(), 2))
print("Mean AQI after :", round(aqi_clean["AQI"].mean(), 2))
```

Observations:
1. The IQR method found the outlier threshold and any values above the upper cap were capped at that boundary. The mean AQI dropped after treatment, which confirms those extreme values were inflating it.
2. The before and after histograms show the change clearly. The after plot has a cleaner right tail. Capping was chosen over deletion because deleting those rows would create gaps in the time series and affect the yearly trend analysis in the next task.

Task 6:
I converted the Date column into datetime format and extracted the year. I then calculated the average AQI for each year and plotted a line chart. A line chart is the clearest way to show a trend over time, even for someone who does not work with data. I marked the most and least polluted years directly on the chart.

```python
aqi_clean["Date"] = pd.to_datetime(aqi_clean["Date"])
aqi_clean["Year"] = aqi_clean["Date"].dt.year

yearly_aqi = aqi_clean.groupby("Year")["AQI"].mean().reset_index()

max_year = yearly_aqi.loc[yearly_aqi["AQI"].idxmax()]
min_year = yearly_aqi.loc[yearly_aqi["AQI"].idxmin()]

plt.figure(figsize=(9, 5))
plt.plot(yearly_aqi["Year"], yearly_aqi["AQI"], marker="o", color="tomato", linewidth=2)
plt.fill_between(yearly_aqi["Year"], yearly_aqi["AQI"], alpha=0.15, color="tomato")

plt.annotate(
    f'Most polluted ({int(max_year["Year"])})',
    xy=(max_year["Year"], max_year["AQI"]),
    xytext=(max_year["Year"] - 0.5, max_year["AQI"] + 8),
    fontsize=9, color="darkred",
    arrowprops=dict(arrowstyle="->", color="darkred")
)

plt.annotate(
    f'Least polluted ({int(min_year["Year"])})',
    xy=(min_year["Year"], min_year["AQI"]),
    xytext=(min_year["Year"] + 0.2, min_year["AQI"] - 15),
    fontsize=9, color="steelblue",
    arrowprops=dict(arrowstyle="->", color="steelblue")
)

plt.title("Average AQI per Year - India (2015 to 2020)")
plt.xlabel("Year")
plt.ylabel("Average AQI")
plt.xticks(yearly_aqi["Year"].astype(int))
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print(yearly_aqi.to_string(index=False))
```

Observation:
The data shows that air quality improved between 2018 and 2020. The worst year in the dataset is 2018 and the best is 2020. The 2020 drop is mostly explained by the COVID-19 lockdown, which reduced traffic and factory activity sharply. So while the trend looks positive, the most recent improvement is driven by an unusual event rather than a policy change alone.

I extracted the month from the Date column and calculated the average AQI for each month across all cities and years. I used a bar chart so it is easy to compare each month directly. I highlighted October, November and December in a different colour to check the NGO's claim about the harvest season.

Task 8:
The AQI dataset has city-level daily records while the crop dataset has district-level yearly records. They cannot be joined directly in that form. To make them compatible, I first extracted the year from the AQI dates, then mapped each city to its state using a lookup dictionary, and aggregated both datasets to the state-year level. For AQI I took the mean per state per year, and for the crop data I summed area and production across all districts and crops for each state and year. Only after this transformation could the two datasets be meaningfully joined.

```python
city_to_state = {
    "Ahmedabad": "Gujarat", "Delhi": "Delhi", "Lucknow": "Uttar Pradesh",
    "Mumbai": "Maharashtra", "Chennai": "Tamil Nadu", "Kolkata": "West Bengal",
    "Bengaluru": "Karnataka", "Hyderabad": "Telangana", "Patna": "Bihar",
    "Gurugram": "Haryana", "Jaipur": "Rajasthan", "Jodhpur": "Rajasthan",
    "Kochi": "Kerala", "Coimbatore": "Tamil Nadu", "Shillong": "Meghalaya",
    "Visakhapatnam": "Andhra Pradesh", "Chandigarh": "Punjab",
    "Amaravati": "Maharashtra", "Amritsar": "Punjab", "Bhopal": "Madhya Pradesh",
    "Brajrajnagar": "Odisha", "Ernakulam": "Kerala", "Guwahati": "Assam",
    "Haldia": "West Bengal", "Nagpur": "Maharashtra", "Navi Mumbai": "Maharashtra",
    "Pune": "Maharashtra", "Siliguri": "West Bengal", "Talcher": "Odisha",
    "Thiruvananthapuram": "Kerala", "Aizawl": "Mizoram"
}

aqi_clean["State"] = aqi_clean["City"].map(city_to_state)
aqi_clean = aqi_clean.dropna(subset=["State"])

aqi_state_year = aqi_clean.groupby(["State", "Year"])["AQI"].mean().reset_index()
aqi_state_year.columns = ["State_Name", "Crop_Year", "Avg_AQI"]

crop_state_year = crop_clean.groupby(["State_Name", "Crop_Year"]).agg(
    Total_Area=("Area", "sum"),
    Total_Production=("Production", "sum")
).reset_index()

merged = pd.merge(aqi_state_year, crop_state_year,
                  on=["State_Name", "Crop_Year"], how="inner")

print(f"Merged dataset shape: {merged.shape}")
display(merged.head(10))

num_cols = ["Avg_AQI", "Total_Area", "Total_Production"]
corr = merged[num_cols].corr()

plt.figure(figsize=(6, 4))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0,
            linewidths=0.5, linecolor="white", square=True)
plt.title("Correlation Matrix - Merged AQI and Crop Data")
plt.tight_layout()
plt.show()
```

Observations:
1. AQI and total production show a weak negative relationship. States with higher average pollution tend to have slightly lower crop output. This is consistent with the idea that pollution affects agriculture, but the relationship is not strong enough to draw conclusions from on its own because other factors like state size and irrigation also play a role.
2. Total area and total production are strongly positively correlated, which makes sense because a bigger farming area generally produces more crop. This also acts as a check that the merge worked correctly and the data is internally consistent.

The central question being tested is whether states with higher pollution have lower crop yields.

Task A:
I identified the three most polluted and three least polluted states in the merged dataset based on their average AQI. I then compared their average annual crop production using a bar chart.

```python
state_aqi = merged.groupby("State_Name")["Avg_AQI"].mean().sort_values(ascending=False)

top_polluted = state_aqi.head(3).index.tolist()
least_polluted = state_aqi.tail(3).index.tolist()
extreme_states = top_polluted + least_polluted

extreme_data = merged[merged["State_Name"].isin(extreme_states)].copy()
extreme_summary = extreme_data.groupby("State_Name").agg(
    Avg_AQI=("Avg_AQI", "mean"),
    Avg_Production=("Total_Production", "mean")
).reset_index()
extreme_summary["Group"] = extreme_summary["State_Name"].apply(
    lambda x: "Most Polluted" if x in top_polluted else "Least Polluted"
)
extreme_summary = extreme_summary.sort_values("Avg_AQI", ascending=False)

colors = ["tomato" if g == "Most Polluted" else "steelblue"
          for g in extreme_summary["Group"]]

plt.figure(figsize=(10, 5))
plt.bar(extreme_summary["State_Name"], extreme_summary["Avg_Production"],
        color=colors, edgecolor="white")
plt.title("Average Crop Production - Most vs Least Polluted States")
plt.xlabel("State")
plt.ylabel("Average Annual Production (tonnes)")
plt.xticks(rotation=15)

from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor="tomato", label="Most polluted"),
    Patch(facecolor="steelblue", label="Least polluted")
]
plt.legend(handles=legend_elements)
plt.tight_layout()
plt.show()

display(extreme_summary[["State_Name","Group","Avg_AQI","Avg_Production"]])
```

Observation:
The pattern is more complicated than the hypothesis would suggest. Some highly polluted states like Delhi have lower crop production, which seems to support the idea. But Delhi is also a heavily urbanised state with very little agricultural land, so the lower production is not necessarily caused by the pollution. Some of the least polluted states also show moderate production, partly because they are smaller states with less farmland overall. The extremes alone do not confirm or disprove the hypothesis.

Task B:
I calculated the Pearson correlation between average state AQI and average crop production and visualised it with a scatter plot and a regression line. This gives the research team both a number and a visual to interpret together.

```python
from scipy import stats

state_summary = merged.groupby("State_Name").agg(
    Avg_AQI=("Avg_AQI", "mean"),
    Avg_Production=("Total_Production", "mean")
).reset_index()

r, p = stats.pearsonr(state_summary["Avg_AQI"], state_summary["Avg_Production"])

plt.figure(figsize=(7, 5))
sns.regplot(data=state_summary, x="Avg_AQI", y="Avg_Production",
            scatter_kws={"color": "steelblue", "alpha": 0.7},
            line_kws={"color": "tomato", "linewidth": 2})

for _, row in state_summary.iterrows():
    plt.text(row["Avg_AQI"] + 0.5, row["Avg_Production"], row["State_Name"],
             fontsize=7, alpha=0.75)

plt.title(f"AQI vs Crop Production by State (r = {r:.2f}, p = {p:.3f})")
plt.xlabel("Average AQI")
plt.ylabel("Average Annual Crop Production (tonnes)")
plt.tight_layout()
plt.show()

print(f"Pearson r = {r:.4f}")
print(f"p-value   = {p:.4f}")
```

Observation:
The Pearson correlation shows a weak to moderate negative relationship between state AQI and crop production. A higher pollution level is linked to slightly lower production, but the r value is low enough that AQI alone explains very little of the variation in production. This means the relationship exists but is not strong. It does not prove that pollution causes lower yields. States differ in rainfall, irrigation, soil type and economic conditions, none of which are in this dataset, and any of these could explain the pattern.
